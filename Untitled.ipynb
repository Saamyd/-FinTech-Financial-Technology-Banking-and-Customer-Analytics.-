```python
# Outliers handling via IQR capping
def cap_outliers_iqr(series, factor= 1.5):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower_limit = max(0.0, q1 - factor * iqr)
    upper_limit = q3 + factor * iqr
    return series.clip(lower = lower_limit, upper = upper_limit)

for col in ['annual_income_inr', 'monthly_spend_inr', 'average_balance_inr']:
    if col in df.columns:
        df[col] = cap_outliers_iqr(df[col])
```

```python
# Column Transformer definition

numeric_feature = ['age', 'annual_income_inr', 'account_tenure_months', 'monthly_transactions', 'monthly_spend_inr', 
                   'average_balance_inr','failed_payments_6m', 'support_tickets_6m', 'days_since_last_login',
                   'spend_to_income_ratio', 'balance_to_spend_ratio', 'friction_score']
category_feature = ['region', 'employment_type', 'app_plan', 'risk_segment',]
pass_feature = ['has_savings_goal', 'auto_save_enabled']
```

# Model Development & Benchmark Evaluation

```python
# Candidate Models

models = {
    'Logistic_Regression': LogisticRegression(
        class_weight= 'balanced', max_iter= 1000, random_state= 89
    ),
    'Random_Forest': RandomForestClassifier(
        n_estimators= 150, max_depth= 12, class_weight= 'balanced', random_state= 89, n_jobs= 1
    ),
    'Extra_Trees': ExtraTreesClassifier(
        n_estimators= 150, max_depth= 12, class_weight= 'balanced', random_state= 89, n_jobs= 1
    )
}
```

```python
from sklearn.impute import SimpleImputer
numeric_transformer = Pipeline(steps= 
                               [('imputer', 
                                 SimpleImputer(strategy= 'most_frequent')),
                                ('encoder',
                                OneHotEncoder(handle_unknown= 'ignore'))
                                ])
```

```python
# Evaluate each model on test data

results = []
fitted_pipelines = {}

for name, clf in models.items():
    pipe= Pipeline(steps= [('preprocessor', preprocessor), ('classifier', clf)])
    pipe.fit(x_train, y_train)
    fitted_pipelines[name]= pipe

    y_pred = pipe.predict(x_test)
    y_prob =pipe.predict_proba(x_test)[:, 1]

    prec_vals, rec_vals, _ = precision_recall_curve(y_test, y_prob)
    pre_auc = auc(rec_vals, prec_vals)
    roc_auc = roc_auc_score(y_test, y_prob)

    results.append({
        'Model' : name,
        'Accuracy' : round(accuracy_score(y_test, y_pred), 4),
        'Precision' : round(precision_score(y_test, y_pred), 4),
        'Recall' : round(recall_score(y_test, y_pred), 4),
        'F1_score' : round(f1_score(y_test, y_pred), 4),
        'ROC_AUC' : round(roc_auc, 4),
        'PRE_AUC' : round(pre_auc, 4)
    })

eval_df = pd.DataFrame(results)
display(eval_df)
```

```python
# Optimize Threshold to Maximize F1-score

f1_scores = (2 * precision * recalls) / (precision + recalls + 1e-9)
best_idx = np.argmax(f1_scores[:-1])
optimal_threshold = float(thresholds[best_idx])
print(f"Optimal Classification Threshold: {optimal_threshold:.4f} (peak F1: {f1_scores[best_idx]:.4f})")
```

```python
# Categorize risk tiers

conditions = [
    (df['churn_probability'] < 0.35),
    (df['churn_probability'] >= 0.35) & (df['churn_probability'] < 0.65),
    (df['churn_probability'] >= 0.65)
]
labels = ['Low Risk', 'Medium Risk', 'High Risk']
df['risk_category'] = np.select(conditions, labels, default= 'Medium Risk')
```

```python
df[['risk_category']]
```

```python
df
```

```python
# Assign Tailored Retention Recommendations

def assign_action(row):
    if row['risk_category'] == 'High Risk':
        if row['friction_score'] >= 2.0:
            return 'Escalate: Priority support call + free refund'
        return 'Relationship Manager Outreach: Personalized saving incentive'
    elif row['risk_category'] == 'Medium Risk':
        if row['auto_save_enabled'] == 0:
            return 'Push Campaign: ₹100 incentive to enable Auto-Save'
        return 'Engagement: 90-day goal challenge notification'
    else:
        return 'Nurture: Cross-cell wealth & investment products'

df['recommended_retention_action'] = df.apply(assign_action, axis= 1)
```
